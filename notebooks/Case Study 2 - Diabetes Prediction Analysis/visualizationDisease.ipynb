```python
import pandas as pd

df = pd.read_csv('cleaned_disease_prediction.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='Age',
    bins=20,
    kde=True
)

plt.title('Age Distribution')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='BMI',
    bins=20,
    kde=True
)

plt.title('BMI Distribution')

plt.show()
```

```python
diabetes_counts = df['Diabetes'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    diabetes_counts,
    labels=['Non-Diabetic','Diabetic'],
    autopct='%1.1f%%'
)

plt.title('Diabetes Distribution')

plt.show()
```

```python
sns.countplot(
    data=df,
    x='Gender',
    hue='Diabetes'
)

plt.title('Diabetes by Gender')

plt.show()
```

```python
sns.countplot(
    data=df,
    x='Blood Pressure',
    hue='Diabetes'
)

plt.title('Diabetes by Blood Pressure')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Diabetes',
    y='BMI'
)

plt.title('BMI by Diabetes Status')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Diabetes',
    y='Age'
)

plt.title('Age by Diabetes Status')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Blood Pressure',
    y='BMI'
)

plt.title('BMI by Blood Pressure')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Age',
    y='BMI',
    hue='Diabetes'
)

plt.title('Age vs BMI')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Age',
    y='BMI',
    hue='Gender',
    style='Diabetes'
)

plt.title('Age vs BMI by Gender')

plt.show()
```

```python
disease_cols = [
    'Heart Disease',
    'Diabetes',
    'Stroke',
    'Kidney Disease',
    'Cancer',
    "Alzheimer's Disease",
    'COPD',
    'Liver Disease',
    "Parkinson's Disease",
    'Tuberculosis',
    'Age',
    'BMI'
]

plt.figure(figsize=(12,8))

sns.heatmap(
    df[disease_cols].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```

```python
age_diabetes = pd.crosstab(
    df['Age Group'],
    df['Diabetes']
)

plt.figure(figsize=(8,5))

sns.heatmap(
    age_diabetes,
    annot=True,
    fmt='d',
    cmap='YlOrRd'
)

plt.title('Age Group vs Diabetes')

plt.show()
```

```python
bmi_diabetes = pd.crosstab(
    df['BMI Category'],
    df['Diabetes']
)

plt.figure(figsize=(8,5))

sns.heatmap(
    bmi_diabetes,
    annot=True,
    fmt='d',
    cmap='Greens'
)

plt.title('BMI Category vs Diabetes')

plt.show()
```

```python
bp_diabetes = pd.crosstab(
    df['Blood Pressure'],
    df['Diabetes']
)

plt.figure(figsize=(8,5))

sns.heatmap(
    bp_diabetes,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title('Blood Pressure vs Diabetes')

plt.show()
```
