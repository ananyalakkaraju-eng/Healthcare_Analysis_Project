```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\Ananya Lakkaraju\Downloads\healthcare_disease_prediction_dataset.csv")
```

```python
print(df.head())

print(df.info())

print(df.shape)

print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print("Duplicates:", df.duplicated().sum())

df.drop_duplicates(inplace=True)
```

```python
text_cols = df.select_dtypes(include='object').columns

for col in text_cols:
    df[col] = df[col].str.strip().str.title()
```

```python
df['Age Group'] = pd.cut(
    df['Age'],
    bins=[0,20,40,60,80,100],
    labels=[
        '0-20',
        '21-40',
        '41-60',
        '61-80',
        '80+'
    ]
)
print(df)
```

```python
df['BMI Category'] = pd.cut(
    df['BMI'],
    bins=[0,18.5,25,30,100],
    labels=[
        'Underweight',
        'Normal',
        'Overweight',
        'Obese'
    ]
)
print(df)
```

```python
df.to_csv(
    "cleaned_disease_prediction.csv",
    index=False
)
```
