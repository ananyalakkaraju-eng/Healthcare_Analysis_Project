```python
import pandas as pd

df = pd.read_csv('cleanedHealthcare_data.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(data=df, x='Gender')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig, ax = plt.subplots(figsize=(10, 6))

counts = df['Medical Condition'].value_counts().head(10)

bars = ax.barh(
    counts.index[::-1],
    counts.values[::-1],
    color='#534AB7',
    edgecolor='none',
    height=0.6
)

# Style the axes
ax.set_xlabel('Patient Count', fontsize=12, color='#5F5E5A', labelpad=10)
ax.set_title('Top 10 Medical Conditions', fontsize=15, fontweight='medium',
             color='#2C2C2A', pad=16)

ax.spines[['top', 'right', 'left']].set_visible(False)
ax.spines['bottom'].set_color('#D3D1C7')

ax.tick_params(axis='both', which='both', length=0)
ax.tick_params(axis='y', labelsize=12, labelcolor='#444441')
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.histplot(df['Billing Amount'], bins=30, kde=True, ax=ax,
             color='#534AB7', edgecolor='white', linewidth=0.4,
             line_kws={'color': '#AFA9EC', 'linewidth': 2})
ax.set_title('Distribution of Billing Amount', fontsize=15, fontweight='medium', color='#2C2C2A', pad=16)
ax.set_xlabel('Billing Amount', fontsize=12, color='#5F5E5A', labelpad=10)
ax.set_ylabel('Frequency', fontsize=12, color='#5F5E5A', labelpad=10)
ax.spines[['top', 'right']].set_visible(False)
ax.spines[['left', 'bottom']].set_color('#D3D1C7')
ax.tick_params(axis='both', length=0, labelsize=11, labelcolor='#5F5E5A')
ax.yaxis.grid(True, color='#D3D1C7', linewidth=0.6, linestyle='--', alpha=0.7)
ax.set_axisbelow(True)
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${int(x):,}'))
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.scatterplot(x='Age', y='Billing Amount', data=df, ax=ax,
                color='#534AB7', alpha=0.45, s=40, edgecolor='none')
ax.set_title('Age vs Billing Amount', fontsize=15, fontweight='medium', color='#2C2C2A', pad=16)
ax.set_xlabel('Age', fontsize=12, color='#5F5E5A', labelpad=10)
ax.set_ylabel('Billing Amount', fontsize=12, color='#5F5E5A', labelpad=10)
ax.spines[['top', 'right']].set_visible(False)
ax.spines[['left', 'bottom']].set_color('#D3D1C7')
ax.tick_params(axis='both', length=0, labelsize=11, labelcolor='#5F5E5A')
ax.yaxis.grid(True, color='#D3D1C7', linewidth=0.6, linestyle='--', alpha=0.7)
ax.set_axisbelow(True)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${int(x):,}'))
plt.tight_layout()
plt.show()
```

```python
gender_counts = df['Gender'].value_counts()
colors = ['#534AB7', '#1D9E75', '#D85A30', '#D4537E', '#378ADD']
fig, ax = plt.subplots(figsize=(6, 6))
wedges, texts, autotexts = ax.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    colors=colors[:len(gender_counts)],
    startangle=90,
    pctdistance=0.75,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}
)
for t in texts:
    t.set_fontsize(12); t.set_color('#444441')
for at in autotexts:
    at.set_fontsize(11); at.set_color('white'); at.set_fontweight('medium')
ax.set_title('Gender Distribution', fontsize=15, fontweight='medium', color='#2C2C2A', pad=16)
plt.tight_layout()
plt.show()
```

```python
insurance_counts = df['Insurance Provider'].value_counts()
colors = ['#534AB7', '#1D9E75', '#D85A30', '#D4537E', '#378ADD', '#BA7517']
fig, ax = plt.subplots(figsize=(7, 7))
wedges, texts, autotexts = ax.pie(
    insurance_counts,
    labels=insurance_counts.index,
    autopct='%1.1f%%',
    colors=colors[:len(insurance_counts)],
    startangle=90,
    pctdistance=0.78,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}
)
for t in texts:
    t.set_fontsize(11); t.set_color('#444441')
for at in autotexts:
    at.set_fontsize(10); at.set_color('white'); at.set_fontweight('medium')
ax.set_title('Insurance Provider Distribution', fontsize=15, fontweight='medium', color='#2C2C2A', pad=16)
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')
corr = numeric_df.corr()
fig, ax = plt.subplots(figsize=(6, 3))
sns.heatmap(corr, annot=True, fmt='.2f', ax=ax,
            cmap=sns.diverging_palette(250, 10, as_cmap=True),
            center=0, linewidths=0.5, linecolor='white',
            annot_kws={'size': 11, 'color': '#2C2C2A'},
            cbar_kws={'shrink': 0.8})
ax.set_title('Correlation Heatmap', fontsize=15, fontweight='medium', color='#2C2C2A', pad=16)
ax.tick_params(axis='both', length=0, labelsize=11, labelcolor='#444441')
plt.tight_layout()
plt.show()
```

```python
df['Age Group'] = pd.cut(df['Age'], bins=[0,18,35,50,65,120],
                          labels=['0-18','19-35','36-50','51-65','65+'])
pivot_table = pd.pivot_table(df, values='Name', index='Medical Condition',
                              columns='Age Group', aggfunc='count', fill_value=0)
fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(pivot_table, annot=True, fmt='d', ax=ax,
            cmap=sns.light_palette('#534AB7', as_cmap=True),
            linewidths=0.5, linecolor='white',
            annot_kws={'size': 11, 'color': '#2C2C2A'},
            cbar_kws={'shrink': 0.7})
ax.set_title('Disease Distribution Across Age Groups', fontsize=15, fontweight='medium', color='#2C2C2A', pad=16)
ax.set_xlabel('Age Group', fontsize=12, color='#5F5E5A', labelpad=10)
ax.set_ylabel('Medical Condition', fontsize=12, color='#5F5E5A', labelpad=10)
ax.tick_params(axis='both', length=0, labelsize=11, labelcolor='#444441')
plt.tight_layout()
plt.show()
```

```python
insurance_heatmap = pd.pivot_table(df, values='Name', index='Insurance Provider',
                                    columns='Medical Condition', aggfunc='count', fill_value=0)
fig, ax = plt.subplots(figsize=(10, 5))
sns.heatmap(insurance_heatmap, annot=True, fmt='d', ax=ax,
            cmap=sns.light_palette('#1D9E75', as_cmap=True),
            linewidths=0.5, linecolor='white',
            annot_kws={'size': 10, 'color': '#2C2C2A'},
            cbar_kws={'shrink': 0.7})
ax.set_title('Insurance Provider vs Medical Condition', fontsize=15, fontweight='medium', color='#2C2C2A', pad=16)
ax.set_xlabel('Medical Condition', fontsize=12, color='#5F5E5A', labelpad=10)
ax.set_ylabel('Insurance Provider', fontsize=12, color='#5F5E5A', labelpad=10)
ax.tick_params(axis='both', length=0, labelsize=11, labelcolor='#444441')
plt.xticks(rotation=35, ha='right')
plt.tight_layout()
plt.show()
```
