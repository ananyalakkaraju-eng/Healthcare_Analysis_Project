```python
import pandas as pd

df = pd.read_csv('cleanedHospital_data.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
status_counts = df['status'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    status_counts,
    labels=status_counts.index,
    autopct='%1.1f%%'
)

plt.title("Appointment Status Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='reason_for_visit'
)

plt.title("Reason For Visit")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x='Month'
)

plt.title("Appointments by Month")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x='Day'
)

plt.title("Appointments by Day")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    y='doctor_id',
    order=df['doctor_id'].value_counts().index
)

plt.title("Appointments Per Doctor")

plt.show()
```

```python
status_reason = pd.crosstab(
    df['reason_for_visit'],
    df['status']
)

status_reason.plot(
    kind='bar',
    stacked=True,
    figsize=(10,6)
)

plt.title("Status by Visit Reason")

plt.show()
```

```python
reason_status = pd.crosstab(
    df['reason_for_visit'],
    df['status']
)

plt.figure(figsize=(8,5))

sns.heatmap(
    reason_status,
    annot=True,
    fmt='d',
    cmap='YlOrRd'
)

plt.title("Visit Reason vs Status")

plt.show()
```

```python
doctor_status = pd.crosstab(
    df['doctor_id'],
    df['status']
)

plt.figure(figsize=(10,6))

sns.heatmap(
    doctor_status,
    annot=True,
    cmap='Blues'
)

plt.title("Doctor vs Appointment Status")

plt.show()
```

```python
doctor_counts = df['doctor_id'].value_counts()

plt.figure(figsize=(8,5))

sns.boxplot(
    x=doctor_counts
)

plt.title("Distribution of Doctor Workload")

plt.show()
```

```python
doctor_summary = (
    df.groupby('doctor_id')
    .size()
    .reset_index(name='Appointments')
)

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=doctor_summary,
    x='doctor_id',
    y='Appointments'
)

plt.title("Doctor Appointment Load")

plt.xticks(rotation=45)

plt.show()
```
