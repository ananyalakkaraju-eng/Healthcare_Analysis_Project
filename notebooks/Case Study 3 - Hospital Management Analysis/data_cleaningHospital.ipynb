```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\Ananya Lakkaraju\Downloads\appointments.csv")
```

```python
print(df.head())

print(df.info())

print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())

df.drop_duplicates(inplace=True)
```

```python
df['appointment_date'] = pd.to_datetime(
    df['appointment_date']
)
```

```python
df['Month'] = df['appointment_date'].dt.month_name()
```

```python
df['Day'] = df['appointment_date'].dt.day_name()
```

```python
text_cols = [
    'reason_for_visit',
    'status'
]

for col in text_cols:
    df[col] = df[col].str.strip().str.title()
```

```python
df.to_csv("cleanedHospital_data.csv", index=False)
```
